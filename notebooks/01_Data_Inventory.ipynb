## Notebook 01 — Data Inventory

### Objective

The objective of this notebook is to perform a complete inventory of all datasets used in the Banking Customer Churn Prediction project.

```python
# ==========================================================
# IMPORT LIBRARIES
# ==========================================================

from pathlib import Path
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.max_colwidth", 120)

print("=" * 60)
print("PFE ESB - Banking Churn Prediction")
print("=" * 60)
print("Pandas :", pd.__version__)
print("NumPy  :", np.__version__)
```

```python
# ==========================================================
# PROJECT PATHS
# ==========================================================

PROJECT_ROOT = Path.cwd().parent

RAW_DATA = PROJECT_ROOT / "data" / "raw"
REFERENCE_DATA = PROJECT_ROOT / "data" / "reference"

OUTPUT_TABLES = PROJECT_ROOT / "outputs" / "tables"

OUTPUT_TABLES.mkdir(parents=True, exist_ok=True)

print("Project root :", PROJECT_ROOT)
print("Raw data     :", RAW_DATA)
print("Reference    :", REFERENCE_DATA)
print("Outputs      :", OUTPUT_TABLES)
```

```python
# ==========================================================
# VERIFY AVAILABLE FILES
# ==========================================================

raw_files = sorted(RAW_DATA.glob("*"))
reference_files = sorted(REFERENCE_DATA.glob("*"))

print("=" * 60)
print("RAW DATA FILES")
print("=" * 60)

for file_path in raw_files:
    print(f"- {file_path.name}")

print("\n" + "=" * 60)
print("REFERENCE DATA FILES")
print("=" * 60)

for file_path in reference_files:
    print(f"- {file_path.name}")
```

```python
# ==========================================================
# GENERIC DATA LOADER
# ==========================================================

def load_dataset(file_path: Path) -> pd.DataFrame:
    """
    Load a CSV or Excel dataset.

    CSV files in this project use ';' as separator.
    Excel files are loaded from their first worksheet.

    Parameters
    ----------
    file_path : Path
        Path of the dataset.

    Returns
    -------
    pd.DataFrame
        Loaded dataset.
    """

    extension = file_path.suffix.lower()

    if extension == ".csv":
        try:
            return pd.read_csv(
                file_path,
                sep=";",
                encoding="utf-8",
                low_memory=False
            )
        except UnicodeDecodeError:
            return pd.read_csv(
                file_path,
                sep=";",
                encoding="latin1",
                low_memory=False
            )

    if extension in [".xlsx", ".xls"]:
        return pd.read_excel(file_path)

    raise ValueError(f"Unsupported file format: {extension}")
```

```python
# ==========================================================
# LOAD ALL DATASETS
# ==========================================================

datasets = {}

all_files = raw_files + reference_files

for file_path in all_files:
    print(f"Loading: {file_path.name}")

    datasets[file_path.name] = load_dataset(file_path)

print("\n" + "=" * 60)
print("All datasets successfully loaded.")
print("=" * 60)
```

```python
# ==========================================================
# DATASET INVENTORY
# ==========================================================

inventory = []

for name, dataframe in datasets.items():

    inventory.append({
        "Dataset": name,
        "Rows": dataframe.shape[0],
        "Columns": dataframe.shape[1],
        "Missing Values": int(dataframe.isna().sum().sum()),
        "Duplicate Rows": int(dataframe.duplicated().sum())
    })

inventory_df = pd.DataFrame(inventory)

display(inventory_df)
```

```python
# ==========================================================
# CHECK MAIN DATASET
# ==========================================================

df_data = datasets["DATA (1).csv"]

print("Shape:", df_data.shape)
print("\nColumns:")
print(df_data.columns.tolist())

display(df_data.head())
```

```python
# ==========================================================
# SAVE INVENTORY
# ==========================================================

inventory_output = OUTPUT_TABLES / "01_dataset_inventory.xlsx"

inventory_df.to_excel(
    inventory_output,
    index=False
)

print("Inventory saved to:")
print(inventory_output)
```

```python
# ==========================================================
# DATA DICTIONARY
# ==========================================================

data_dictionary = pd.DataFrame({

    "Column": df_data.columns,

    "Data Type": df_data.dtypes.astype(str),

    "Non Missing Values": df_data.notna().sum().values,

    "Missing Values": df_data.isna().sum().values,

    "Missing (%)":
        (df_data.isna().mean()*100).round(2).values,

    "Unique Values":
        df_data.nunique(dropna=True).values

})

display(data_dictionary)
```

```python
# ==========================================================
# SAVE DATA DICTIONARY
# ==========================================================

data_dictionary.to_excel(

    OUTPUT_TABLES /

    "02_DATA_Dictionary.xlsx",

    index=False

)

print("DATA Dictionary exported successfully.")
```
