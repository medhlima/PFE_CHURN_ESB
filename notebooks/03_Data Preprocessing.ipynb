## 3.4 Data Type Correction

The raw dataset contains identifiers, categorical variables, numerical values and date fields stored as numeric values.

This step aims to:

- preserve identifiers as strings;
- convert date fields into datetime format;
- convert categorical variables into clean string values;
- keep financial variables as numerical values;
- detect invalid date formats without modifying the original source file.

```python
# ==========================================================
# COLUMN GROUPS
# ==========================================================

identifier_columns = [
    "CUSTOMER_NO",
    "ACCOUNT_NO"
]

standard_date_columns = [
    "CUST_OPENING_DATE",
    "DATE_OF_BIRTH",
    "LAST_REVIEW_DATE",
    "NEXT__REVIEW_DATE",
    "ACCT_OPENING_DATE",
    "ACCT_CLOSE_DATE"
]

special_date_columns = [
    "STARTDATE",
    "MATURITYDATE"
]

categorical_columns = [
    "NATIONALITY",
    "RESIDENCE",
    "MARITAL_STATUS",
    "NATURE_CLIENT",
    "SCORE_KYC",
    "COMPLETED_FILE",
    "ACCOUNT_STATUS",
    "ACCOUNT_TYPE_DESC",
    "CURRENCY",
    "CLOSURE_REASON",
    "PRODUCT_GROUP",
    "PRODUCT_LINE",
    "PRODUCT",
    "ACCOUNTNATURE",
    "PRODUCT_STATUS",
    "PARTYCLASS"
]

numeric_columns = [
    "ACCT_BALANCE",
    "SALARY",
    "AMOUNT",
    "FIXEDRATE"
]

code_columns = [
    "BRANCH",
    "INDUSTRY",
    "LOB",
    "ACCOUNT_CATEGORY"
]

print("Identifiers          :", identifier_columns)
print("Standard dates       :", standard_date_columns)
print("Special-format dates :", special_date_columns)
print("Categorical          :", len(categorical_columns))
print("Numeric              :", numeric_columns)
print("Codes                :", code_columns)
```

```python
# ==========================================================
# WORKING COPY
# ==========================================================

df_clean = df.copy()

print("Original dataset :", df.shape)
print("Working copy     :", df_clean.shape)
```

```python
# ==========================================================
# IDENTIFIER CONVERSION
# ==========================================================

for column in identifier_columns:
    df_clean[column] = (
        pd.to_numeric(df_clean[column], errors="coerce")
          .astype("Int64")
          .astype("string")
    )

display(df_clean[identifier_columns].head())

print(df_clean[identifier_columns].dtypes)
```

```python
# ==========================================================
# SAFE YYYYMMDD DATE CONVERSION
# ==========================================================

def convert_yyyymmdd(
    series: pd.Series,
    min_year: int = 1900,
    max_year: int = 2100
) -> pd.Series:
    """
    Convert standard YYYYMMDD values into pandas datetime.

    Invalid formats, impossible calendar dates and years outside
    the accepted interval are converted to NaT.
    """

    cleaned = (
        series
        .astype("string")
        .str.replace(r"\.0$", "", regex=True)
        .str.strip()
    )

    valid_format = cleaned.str.fullmatch(r"\d{8}", na=False)

    years = pd.to_numeric(
        cleaned.str.slice(0, 4),
        errors="coerce"
    )

    valid_year = years.between(
        min_year,
        max_year,
        inclusive="both"
    )

    eligible = valid_format & valid_year
    safe_values = cleaned.where(eligible)

    return pd.to_datetime(
        safe_values,
        format="%Y%m%d",
        errors="coerce"
    )
```

```python
# ==========================================================
# STANDARD DATE CONVERSION
# ==========================================================

date_conversion_report = []

for column in standard_date_columns:

    original_non_null = int(
        df_clean[column].notna().sum()
    )

    converted = convert_yyyymmdd(
        df_clean[column],
        min_year=1900,
        max_year=2100
    )

    valid_dates = int(converted.notna().sum())
    invalid_dates = original_non_null - valid_dates

    df_clean[column] = converted

    invalid_rate = (
        invalid_dates / original_non_null * 100
        if original_non_null > 0
        else 0
    )

    date_conversion_report.append({
        "Column": column,
        "Original Non-Null": original_non_null,
        "Valid Dates": valid_dates,
        "Invalid Dates Converted to NaT": invalid_dates,
        "Invalid Rate (%)": round(invalid_rate, 2)
    })

date_conversion_report = pd.DataFrame(
    date_conversion_report
)

display(date_conversion_report)
```

## 3.5 Missing Values Treatment

Missing values are treated according to their business meaning.

The strategy distinguishes between:

- structural missing values;
- missing customer information;
- missing account information;
- missing numerical values;
- missing categorical values.

No global deletion rule is applied.

```python
# ==========================================================
# MISSING VALUES BEFORE TREATMENT
# ==========================================================

missing_before = pd.DataFrame({
    "Column": df_clean.columns,
    "Missing Before": df_clean.isna().sum().values,
    "Missing Before (%)": (
        df_clean.isna().mean() * 100
    ).round(2).values
})

display(
    missing_before.sort_values(
        "Missing Before (%)",
        ascending=False
    )
)
```

```python
# ==========================================================
# MISSING VALUE STRATEGY
# ==========================================================

categorical_fill_unknown = [
    "MARITAL_STATUS",
    "NATURE_CLIENT",
    "SCORE_KYC",
    "COMPLETED_FILE"
]

structural_fill_not_applicable = [
    "ACCOUNT_STATUS",
    "ACCOUNT_TYPE_DESC",
    "CURRENCY",
    "CLOSURE_REASON",
    "PRODUCT_GROUP",
    "PRODUCT_LINE",
    "PRODUCT",
    "ACCOUNTNATURE",
    "PRODUCT_STATUS"
]

numeric_fill_median = [
    "SALARY"
]

numeric_keep_missing = [
    "ACCT_BALANCE",
    "AMOUNT",
    "FIXEDRATE"
]

print("Unknown category fill:", categorical_fill_unknown)
print("Not applicable fill:", structural_fill_not_applicable)
print("Median fill:", numeric_fill_median)
print("Keep missing:", numeric_keep_missing)
```

```python
# ==========================================================
# APPLY MISSING VALUE TREATMENT
# ==========================================================

for column in categorical_fill_unknown:
    df_clean[column] = (
        df_clean[column]
        .astype("string")
        .fillna("UNKNOWN")
    )

for column in structural_fill_not_applicable:
    df_clean[column] = (
        df_clean[column]
        .astype("string")
        .fillna("NOT_APPLICABLE")
    )

for column in numeric_fill_median:
    median_value = df_clean[column].median()
    df_clean[column] = df_clean[column].fillna(median_value)

print("Missing value treatment applied.")
```

```python
# ==========================================================
# MISSING VALUES AFTER TREATMENT
# ==========================================================

missing_after = pd.DataFrame({
    "Column": df_clean.columns,
    "Missing After": df_clean.isna().sum().values,
    "Missing After (%)": (
        df_clean.isna().mean() * 100
    ).round(2).values
})

missing_comparison = missing_before.merge(
    missing_after,
    on="Column",
    how="left"
)

display(
    missing_comparison.sort_values(
        "Missing Before (%)",
        ascending=False
    )
)
```

## 3.6 Exact Duplicate Removal

Exact duplicate rows are removed to eliminate redundant records while preserving unique customer-account information.

Only fully duplicated rows are removed.

```python
# ==========================================================
# EXACT DUPLICATE REMOVAL
# ==========================================================

rows_before = len(df_clean)

duplicates = df_clean.duplicated().sum()

df_clean = df_clean.drop_duplicates()

rows_after = len(df_clean)

removed = rows_before - rows_after

print("=" * 70)
print("EXACT DUPLICATE REMOVAL")
print("=" * 70)

print(f"Rows before     : {rows_before:,}")
print(f"Duplicate rows  : {duplicates:,}")
print(f"Rows after      : {rows_after:,}")
print(f"Removed rows    : {removed:,}")
```

```python
remaining_duplicates = df_clean.duplicated().sum()

print()

print("Remaining duplicates :", remaining_duplicates)
```
