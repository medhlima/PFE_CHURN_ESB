```python
from pathlib import Path

import pandas as pd

# ==========================================================
# PIPELINE V2 FINAL VALIDATION
# ==========================================================

PROJECT_ROOT = Path.cwd().parent

MODEL_FILE = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "dataset_modelisation.parquet"
)

TARGET_FILE = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "cible_client.parquet"
)

if not MODEL_FILE.exists():
    raise FileNotFoundError(
        f"Model dataset not found: {MODEL_FILE}"
    )

if not TARGET_FILE.exists():
    raise FileNotFoundError(
        f"Target dataset not found: {TARGET_FILE}"
    )

df = pd.read_parquet(MODEL_FILE)
target_df = pd.read_parquet(TARGET_FILE)

print("=" * 70)
print("PIPELINE V2 FINAL VALIDATION")
print("=" * 70)

print(f"Model file       : {MODEL_FILE}")
print(f"Rows             : {len(df):,}")
print(f"Columns          : {df.shape[1]}")
print(f"Unique customers : {df['CUSTOMER_NO'].nunique():,}")
print(f"Duplicate clients: {df['CUSTOMER_NO'].duplicated().sum():,}")
print(f"Missing customer : {df['CUSTOMER_NO'].isna().sum():,}")
print(f"Missing target   : {df['CHURN'].isna().sum():,}")
print(f"Churners         : {df['CHURN'].sum():,}")
print(f"Non-churners     : {(df['CHURN'] == 0).sum():,}")
print(f"Churn rate       : {df['CHURN'].mean() * 100:.2f}%")

leakage_columns = [
    "ACCOUNT_STATUS",
    "ACCT_CLOSE_DATE",
    "CLOSURE_REASON",
    "PRODUCT_STATUS",
    "HAS_ACTIVE_ACCOUNT",
    "TARGET_CHURN",
]

present_leakage = [
    column
    for column in leakage_columns
    if column in df.columns
]

print(f"Leakage columns  : {present_leakage}")

# ==========================================================
# ASSERTIONS
# ==========================================================

assert len(df) == 195_120
assert df["CUSTOMER_NO"].nunique() == 195_120
assert df["CUSTOMER_NO"].duplicated().sum() == 0
assert df["CUSTOMER_NO"].isna().sum() == 0
assert df["CHURN"].isna().sum() == 0
assert set(df["CHURN"].unique()).issubset({0, 1})
assert int(df["CHURN"].sum()) == 10_548
assert round(df["CHURN"].mean() * 100, 2) == 5.41
assert len(present_leakage) == 0

print("\nPipeline V2 validated successfully.")

display(df.head())
display(df.dtypes.to_frame("DATA_TYPE"))
```
